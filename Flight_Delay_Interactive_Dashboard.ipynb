```python
import pandas as pd
import numpy as np
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt
import folium
```

```python
df = pd.read_csv('/content/Airline_Delay_Cause.csv')   # update filename if needed
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Date'] = pd.to_datetime(df['year'].astype(str) + '-' + df['month'].astype(str))
df = df.dropna(subset=['arr_delay'])
df['Month'] = df['Date'].dt.month_name()
```

```python
avg_delay = round(df['arr_delay'].mean(), 2)
cancel_rate = round(df['arr_cancelled'].mean() * 100, 2)

print(f"✈️ Average Arrival Delay: {avg_delay} mins")
print(f"🚫 Cancelled Flights: {cancel_rate}%")
```

```python
fig = px.bar(df.groupby('carrier_name')['arr_delay'].mean().reset_index(),
             x='carrier_name', y='arr_delay', color='carrier_name',
             title='Average Arrival Delay by Airline')
fig.show()
```

```python
fig = px.line(df.groupby('Month')['arr_delay'].mean().reset_index(),
              x='Month', y='arr_delay', markers=True,
              title='Monthly Average Arrival Delay')
fig.show()
```

```python
fig = px.pie(df, names='carrier_name', values='arr_cancelled',
             title='Percentage of Flights Cancelled by Airline')
fig.show()
```

```python
!wget https://raw.githubusercontent.com/jpatokal/openflights/master/data/airports.dat -O airports.csv

import pandas as pd

# Columns in the OpenFlights data
cols = ['AirportID', 'Name', 'City', 'Country', 'IATA', 'ICAO',
        'Latitude', 'Longitude', 'Altitude', 'Timezone', 'DST', 'TzDB', 'Type', 'Source']

airports = pd.read_csv('airports.csv', names=cols)

# Keep only necessary columns
airports = airports[['IATA', 'Latitude', 'Longitude']]
airports.rename(columns={'IATA': 'airport'}, inplace=True)
airports.head()
```

```python
# Assume your flight delay DataFrame is called df
df = df.merge(airports, on='airport', how='left')
df.head()
```

```python
df[['airport', 'airport_name', 'Latitude', 'Longitude', 'arr_delay']].head()
```

```python
import folium

# Base map centered on USA
m = folium.Map(location=[39.8283, -98.5795], zoom_start=4)

# Plot sample of 500 airports to speed up rendering
for _, row in df.dropna(subset=['Latitude', 'Longitude']).sample(500).iterrows():
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=4,
        color='red' if row['arr_delay'] > 0 else 'green',
        fill=True,
        fill_opacity=0.6,
        popup=f"{row['airport_name']}<br>Avg Delay: {row['arr_delay']:.2f} mins"
    ).add_to(m)

m
```
